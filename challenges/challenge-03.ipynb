*Adapted from: Hill, R. C., Griffiths, W. E., & Lim, G. C. (2018). Principles of Econometrics (5th ed.). John Wiley & Sons.*

## Dataset Description

The file `collegetown` contains data on 500 single-family houses sold in Baton Rouge, Louisiana during 2009–2013. The dataset includes the following variables:

- **PRICE**: Selling price in thousands of dollars
- **SQFT**: Size of the house in hundreds of square feet  
- **AGE**: Age of the house measured as a categorical variable (1 = newest, 11 = oldest)

Let X denote all observations on SQFT and AGE.

## Exercise: Multiple Regression Model Estimation

Estimate the following multiple linear regression models using all observations in the dataset:


#### Model 1: Interactive Effects Model
$$PRICE = \beta_1 + \beta_2 SQFT + \beta_3 (SQFT \times AGE) + e$$

#### Model 2: Log-Linear with Square Root Term
$$\ln(PRICE) = \beta_1 + \beta_2 SQFT + \beta_3 SQFT^{1/2} + e$$

#### Model 3: Quadratic Model with Age
$$PRICE = \beta_1 + \beta_2 SQFT + \beta_3 SQFT^2 + \beta_4 AGE + e$$

#### Model 4: Log-Log Model
$$\ln(PRICE) = \beta_1 + \beta_2 \ln(SQFT) + \beta_3 AGE + e$$

#### Model 5: Polynomial with Interaction
$$PRICE = \beta_1 + \beta_2 SQFT + \beta_3 AGE + \beta_4 AGE^2 + \beta_5 (SQFT \times AGE) + e$$

#### Model 6: Semi-Log with Quadratic Age
$$\ln(PRICE) = \beta_1 + \beta_2 SQFT + \beta_3 AGE + \beta_4 AGE^2 + e$$

## Tasks

For each model:

### Part (a)
Estimate the regression model and report the results. Include coefficient estimates, standard errors, t-statistics, and p-values.

### Part (b)
Test whether the estimated coefficients are significantly different from zero at the 5% significance level.

### Part (c)
Report the R-squared and adjusted R-squared for each model.

### Part (d)
Calculate the F-statistic for the overall significance of each regression model.
