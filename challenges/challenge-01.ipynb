## Dataset Description

The data file `collegetown` contains observations on 500 single-family houses sold in Baton Rouge, Louisiana, during 2009–2013. The dataset includes the following variables:

- **PRICE**: Sale price in thousands of dollars
- **SQFT**: Total interior area of the house in hundreds of square feet

## Exercise Questions

### Part (a)
Plot house price against house size in a scatter diagram.

### Part (b)
Estimate the linear regression model:

$$
PRICE = \beta_1 + \beta_2 SQFT + \varepsilon
$$

Interpret the estimates and draw a sketch of the fitted line.

### Part (c)
Estimate the quadratic regression model:

$$
PRICE = \alpha_1 + \alpha_2 SQFT^2 + \varepsilon
$$


### Part (d)
Graph the fitted curve for the model in part (c).

### Part (e)
For the regressions in parts (b) and (c), compute the least squares residuals and plot them against SQFT.

### Part (f)
One basis for choosing between these two specifications is how well the data are fit by the model. Compare the sum of squared residuals (SSE) from the models in parts (b) and (c). Which model has a lower SSE? How does having a lower SSE indicate a "better-fitting" model?
